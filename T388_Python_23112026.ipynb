### WAP to print whether a number is an Armstrong number

```python
num=int(input("Enter any number to check whether it is Armstrong or not:"))

length=len(str(num))
result=0
for i in str(num):
    result= result+(int(i)**length)

if result== num:
    print("This number",num,"is an Armstron Number!")
else:
    print("Oh!, This number is NOT an Armstrong Number.")
```

## Fatorial or Fibonacci Series
