```python
# WAP to check the greatest of the 3 numbers

x=int(input("Please enter  your number1 to check"))
y=int(input("Please enter  your number2 to check"))
z=int(input("Please enter  your number3 to check"))
if x>y>z or x>z>y:
    print(x,"is Greater")
elif y>x>z or y>z>x:
    print(y,"is Greater")

else:
    print(z,"is Greater")
print ("Program Ends here!")
```

```python
# WAP to check the greatest of the 3 numbers

x=int(input("Please enter  your number1 to check"))
y=int(input("Please enter  your number2 to check"))
z=int(input("Please enter  your number3 to check"))
if x>y and x>z:
    print(x,"is Greater")
elif y>x and y>z:
    print(y,"is Greater")

else:
    print(z,"is Greater")
print ("Program Ends here!")
```

```python
# WAP to check the greatest of the 3 numbers

x=int(input("Please enter  your number1 to check"))
y=int(input("Please enter  your number2 to check"))
z=int(input("Please enter  your number3 to check"))
if (x>y) & (x>z):
    print(x,"is Greater")
elif (y>x) & (y>z):
    print(y,"is Greater")

else:
    print(z,"is Greater")
print ("Program Ends here!")
```

### For Loop

```python
# WAP to print 1 to 5
for i in range (1,6): #range (Start,Stop)
    print(i)
```

```python
# WAP to print 1 to 5
for i in range (1,6,1): # range(Start,stop,step)
    print(i)
```

```python
# WAP to print 1 to 50 in multiples of 5

for i in range (0,50,5):
    print(i)
```
